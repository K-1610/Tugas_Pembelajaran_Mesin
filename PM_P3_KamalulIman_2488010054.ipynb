## Latihan Mandiri
**Latihan 1.** Muat dataset lain bawaan seaborn, yaitu `tips` (`sns.load_dataset('tips')`). Tampilkan `head()`, `shape`, dan `info()`.


```python
# TULIS KODE ANDA DI SINI (Latihan 1)

import pandas as pd, seaborn as sns
import matplotlib.pyplot as plt

tips = sns.load_dataset('tips')
display(tips.head())
print(tips.shape)
tips.info()
```

**Latihan 2.** Pada dataset `tips`, buat histogram kolom `total_bill` dan box plot kolom `tip`. Amati sebaran dan outlier-nya.


```python
# TULIS KODE ANDA DI SINI (Latihan 2)

plt.figure(figsize=(10, 5))
sns.histplot(tips['total_bill'], bins=20, kde=True, color='skyblue')
plt.title('Histogram Sebaran Total Bill')
plt.xlabel('Total Bill')
plt.ylabel('Frekuensi')
plt.show()
plt.figure(figsize=(6, 3))
sns.boxplot(x=tips['tip'], color='lightgreen')
plt.title('Box Plot Tip (Deteksi Outlier)')
plt.xlabel('Tip')
plt.show()
```

Histogram total_bill menunjukkan sebaran yang right-skewed (cenderung menceng ke kanan), di mana sebagian besar total tagihan berada di kisaran angka 10 hingga 20. Sementara itu, box plot pada kolom tip mendeteksi adanya beberapa titik di luar batas atas (outlier) yang menunjukkan adanya nilai tip yang relatif sangat tinggi dibandingkan rata-rata umum.

**Latihan 3.** Buat scatter plot `total_bill` (x) vs `tip` (y) pada dataset `tips`, diwarnai menurut kolom `time` (Lunch/Dinner). Apa yang dapat Anda simpulkan?


```python
# TULIS KODE ANDA DI SINI (Latihan 3)

plt.figure(figsize=(8, 4))
sns.scatterplot(data=tips, x='total_bill', y='tip', hue='time', palette='Set2')
plt.title('Scatter Plot: Total Bill vs Tip Berdasarkan Waktu (Time)')
plt.xlabel('Total Bill')
plt.ylabel('Tip')
plt.show()
```

Terdapat korelasi positif yang jelas antara total_bill dan tip—semakin besar total tagihan (total_bill), maka nominal tip yang diberikan cenderung semakin tinggi. Selain itu, dari pembeda warna (time), terlihat bahwa sebagian besar transaksi dan tagihan dengan nominal besar terjadi pada waktu Dinner (makan malam) dibanding Lunch (makan siang).

**Latihan 4.** Buat heatmap korelasi untuk fitur numerik pada dataset `tips`. Pasangan fitur mana yang paling berkorelasi?


```python
# TULIS KODE ANDA DI SINI (Latihan 4)

plt.figure(figsize=(6, 4))
num_tips = tips.select_dtypes(include='number')
sns.heatmap(num_tips.corr(), annot=True, cmap='coolwarm', fmt='.2f', linewidths=0.5)
plt.title('Heatmap Korelasi Fitur Numerik - Dataset Tips')
plt.show()
```

Pasangan fitur numerik yang memiliki korelasi paling kuat adalah total_bill dan tip (atau total_bill dengan size), yang ditunjukkan dengan nilai koefisien korelasi positif yang paling tinggi di antara kombinasi variabel numerik lainnya pada dataset tersebut.
